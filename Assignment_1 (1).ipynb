```python
while True:
    print("\n========== SMART CALCULATOR ==========")

    #  Take Two Numbers As Input
    try:
        num1 = float(input("Enter First Number: "))
        num2 = float(input("Enter Second Number: "))
    except ValueError:
        print("Please enter valid numbers.")
        continue

    # Menu Options:
    print("\n1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Modulus")
    print("6. Power")
    print("7. Floor Division")

    choice = input("Enter Choice (1-7): ")
    result = None

    # Operations Logic
    if choice == "1":
        result = num1 + num2
    elif choice == "2":
        result = num1 - num2
    elif choice == "3":
        result = num1 * num2
    elif choice == "4":
        if num2 == 0:
            print("Cannot divide by zero.")
        else:
            result = num1 / num2
    elif choice == "5":
        if num2 == 0:
            print("Cannot divide by zero.")
        else:
            result = num1 % num2
    elif choice == "6":
        result = num1 ** num2
    elif choice == "7":
        if num2 == 0:
            # Division by zero
            print("Cannot divide by zero.")
        else:
            result = num1 // num2
    else:
        # invaild choice
        print("Invalid Choice")

    # (Positive/Negative/Zero)
    if result is not None:
        print("Answer =", result)
        if result > 0:
            print("The result is Positive.")
        elif result < 0:
            print("The result is Negative.")
        else:
            print("The result is Zero.")

    # Ask the user if they want to calculate again.
    again = input("\nDo you want to calculate again? (yes/no): ").strip().lower()

    if again != "yes" and again != "y":
        print("Thank you for using Smart Calculator!")
        break
```

```python
# Day 3: Input, Operators & Conditions

# 1. Calculator
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operator = input("Enter operator (+, -, *, /): ")

if operator == '+':
    print("Result:", num1 + num2)
elif operator == '-':
    print("Result:", num1 - num2)
elif operator == '*':
    print("Result:", num1 * num2)
elif operator == '/':
    if num2 != 0:
        print("Result:", num1 / num2)
    else:
        print("Error: Division by zero is not allowed.")
else:
    print("Invalid operator entered.")

# 2. Grade Calculator
marks = float(input("\nEnter student marks (0-100): "))
if marks >= 90:
    grade = 'A+'
elif marks >= 80:
    grade = 'A'
elif marks >= 70:
    grade = 'B'
elif marks >= 60:
    grade = 'C'
elif marks >= 50:
    grade = 'D'
else:
    grade = 'F (Fail)'
print("Calculated Grade:", grade)

# 3. Voting Eligibility
age = int(input("\nEnter your age: "))
if age >= 18:
    print("You are eligible to vote!")
else:
    print("You are not eligible to vote yet.")

# 4. Largest of Three Numbers
a = float(input("\nEnter first number: "))
b = float(input("Enter second number: "))
c = float(input("Enter third number: "))

if a >= b and a >= c:
    largest = a
elif b >= a and b >= c:
    largest = b
else:
    largest = c
print("The largest number is:", largest)

# 5. Electricity Bill Calculator
units = float(input("\nEnter electricity units consumed: "))
bill_amount = 0

if units <= 100:
    bill_amount = units * 5
elif units <= 200:
    bill_amount = (100 * 5) + ((units - 100) * 7)
else:
    bill_amount = (100 * 5) + (100 * 7) + ((units - 200) * 10)

print(f"Total Electricity Bill: Rs. {bill_amount:.2f}")
```

```python
# Day 4: Loops Practice

# 1. Multiplication Table
table_num = int(input("Enter number for multiplication table: "))
print(f"--- Multiplication Table of {table_num} ---")
for i in range(1, 11):
    print(f"{table_num} x {i} = {table_num * i}")

# 2. Factorial
n = int(input("\nEnter a number to calculate factorial: "))
factorial = 1
for i in range(1, n + 1):
    factorial *= i
print(f"Factorial of {n} is: {factorial}")

# 3. Sum of N Numbers
limit = int(input("\nEnter value for N to find sum (1 to N): "))
total_sum = 0
for i in range(1, limit + 1):
    total_sum += i
print(f"Sum of numbers from 1 to {limit}: {total_sum}")

# 4. Fibonacci Series
terms = int(input("\nEnter number of Fibonacci terms needed: "))
first, second = 0, 1
print("Fibonacci Series:")
for _ in range(terms):
    print(first, end=" ")
    first, second = second, first + second
print()

# 5. Prime Numbers Check
check_num = int(input("\nEnter a number to check if it's Prime: "))
if check_num > 1:
    is_prime = True
    for i in range(2, int(check_num ** 0.5) + 1):
        if check_num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(f"{check_num} is a Prime Number.")
    else:
        print(f"{check_num} is NOT a Prime Number.")
else:
    print(f"{check_num} is NOT a Prime Number.")
```

```python
# Day 5: Functions

import math

# 1. Even/Odd Function
def check_even_odd(number):
    if number % 2 == 0:
        return f"{number} is Even"
    return f"{number} is Odd"

print(check_even_odd(15))

# 2. Area of Circle
def calculate_circle_area(radius):
    return math.pi * (radius ** 2)

r = 7.0
print(f"Area of circle with radius {r}: {calculate_circle_area(r):.2f}")

# 3. Average Marks
def get_average(marks_list):
    if len(marks_list) == 0:
        return 0
    return sum(marks_list) / len(marks_list)

student_marks = [78, 85, 92, 67, 88]
print("Average Marks:", get_average(student_marks))

# 4. Student Result System
def process_student_result(marks_dict):
    marks = list(marks_dict.values())
    avg = sum(marks) / len(marks)
    status = "Pass" if avg >= 40 else "Fail"
    return {
        "Average": round(avg, 2),
        "Status": status
    }

report_card = {"Maths": 85, "Science": 72, "English": 68}
print("Student Result:", process_student_result(report_card))

# 5. Calculator using Functions
def add(a, b): return a + b
def subtract(a, b): return a - b
def multiply(a, b): return a * b
def divide(a, b): return a / b if b != 0 else "Cannot divide by zero"

print("Addition (10 + 5):", add(10, 5))
print("Division (20 / 4):", divide(20, 4))
```

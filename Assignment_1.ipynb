Part A: Strings

```python
# String with full name

full_name = ("Simran Lubana")
print(full_name)

# Length of the string
print(len(full_name))

# first and last character
print(full_name[0],full_name[-1])

# print the first 5 and last 5 character
print(full_name[0:5])
print(full_name[8:13])

# Print every alternate character using slicing
print(full_name[::2])

# Reverse the string using slicing
print(full_name[::-1])

# String methods (UPPERCASE,LOWERCASE,TITLECASE,CAPITALIZE)

print(full_name.upper())

print(full_name.lower())

print(full_name.title())

print(full_name.capitalize())

# Replace first name with another name

print(full_name.replace("Simran","Virender"))

# Position of word

print(full_name.find("Lubana"))

# Count
print(full_name.count("a"))

# Starts and ends

print(full_name.startswith("S"))

print(full_name.endswith("a"))

# Split the sentance

print(full_name.split())

print(full_name.center(60))

print(full_name.rjust(40))

print(full_name.strip())

```

PART: B LISTS

```python
fruits=["Mango","Lichi","Apple","Date","Pineapple"]

print(" full list:",fruits)

print("first fruit:", fruits[0])

print("last  fruit:", fruits[-1])

# first three fruits

print(fruits[0:3])

# INSERT

fruits.insert(2,"Fig")

print("insert:",fruits)

# Add three fruits using Extend()

fruits.extend(["Grapes","kiwi","Banana"])

print("extend:",fruits)

# Remove

fruits.remove("Banana")
print("remove:",fruits)

# POP

fruits.pop()

print(fruits)

# REPLACE


#sort

fruits.sort()

print(fruits)

# Reverse

fruits.reverse()

print(fruits)

# Copy

new_fruitslist= fruits.copy()
print(new_fruitslist)
```

PROJECT: STUDENT DATABASE

```python
list=["Drishti","komal","Sunaina","Taniya","Payal","Amit","Sahil","Rahul","Anil","kartik"]

# The Complete List

print("Students_List:",list)

# First Student

print("First Student:",list[0])

# Last Student

print("Last Student:",list[9])

# Add a new student

list.append("Nikhil")

print("append():",list)

# Insert a student at index 3

list.insert(3,"Priya")

print("insert():",list)

# Remove one student

list.remove("Sahil")
print("remove():",list)

# Sort the list

list.sort()
print(list)

# Reverse the list

list.reverse()
print(list)

# Copy

new_list=list.copy()

print("new_list:",list)

```

```python
# 6. Grade Calculator using if-elif
score = float(input("Enter score (0-100): "))
if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
elif score >= 60:
    print("Grade: D")
elif score >= 0:
    print("Grade: F")
else:
    print("Invalid Score")

# 7. Voting Eligibility
age = int(input("Enter your age: "))
if age >= 18:
    print("Eligible to vote")
else:
    print("Not eligible to vote")

# 8. Login System
username = input("Username: ")
password = input("Password: ")
if username == "admin" and password == "1234":
    print("Login Successful!")
else:
    print("Invalid Credentials!")

# 9. Leap Year Checker
year = int(input("Enter a year: "))
if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print(f"{year} is a Leap Year")
else:
    print(f"{year} is NOT a Leap Year")

# 10. Calculator using if-elif
num1 = float(input("Enter first number: "))
op = input("Enter operator (+, -, *, /): ")
num2 = float(input("Enter second number: "))

if op == '+':
    print("Result:", num1 + num2)
elif op == '-':
    print("Result:", num1 - num2)
elif op == '*':
    print("Result:", num1 * num2)
elif op == '/':
    if num2 != 0:
        print("Result:", num1 / num2)
    else:
        print("Error: Division by zero!")
else:
    print("Invalid operator!")
```

Part:B Loops

```python
# 11. Print 1 to 20
for i in range(1, 21):
    print(i, end=" ")
print()

# 12. Print Even Numbers (1 to 20)
for i in range(2, 21, 2):
    print(i, end=" ")
print()

# 13. Print Odd Numbers (1 to 20)
for i in range(1, 20, 2):
    print(i, end=" ")
print()

# 14. Multiplication Table
num = int(input("Enter number for table: "))
for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")

# 15. Sum of First N Numbers
n = int(input("Enter N: "))
total_sum = 0
for i in range(1, n + 1):
    total_sum += i
print(f"Sum of first {n} numbers is: {total_sum}")

# 16. Factorial
n = int(input("Enter a number: "))
fact = 1
for i in range(1, n + 1):
    fact *= i
print(f"Factorial of {n} is: {fact}")

# 17. Reverse Counting (10 to 1)
for i in range(10, 0, -1):
    print(i, end=" ")
print()

# 18. Print Characters of a String
text = input("Enter a string: ")
for char in text:
    print(char)

# 19. Print List Elements
my_list = [10, "Python", 3.14, True, "Loop"]
for item in my_list:
    print(item)

# 20. Count Vowels
text = input("Enter a string: ").lower()
vowels = "aeiou"
count = 0
for char in text:
    if char in vowels:
        count += 1
print(f"Number of vowels: {count}")
```

break & continue

```python
# 21. Stop printing when number becomes 6 using break
for i in range(1, 11):
    if i == 6:
        break
    print(i, end=" ")
print()

# 22. Skip number 5 using continue
for i in range(1, 11):
    if i == 5:
        continue
    print(i, end=" ")
print()
```

```python
# 23. Number Guessing Game
import random

target = random.randint(1, 100)
print("--- Number Guessing Game (1 to 100) ---")
while True:
    guess = int(input("Guess the number: "))
    if guess < target:
        print("Too Low!")
    elif guess > target:
        print("Too High!")
    else:
        print("Congratulations! You guessed it correctly!")
        break

# 24. Password Checker
stored_password = "pythonSecret123"
attempts = 3

while attempts > 0:
    pwd = input("Enter password: ")
    if pwd == stored_password:
        print("Access Granted!")
        break
    else:
        attempts -= 1
        print(f"Incorrect! Attempts left: {attempts}")
if attempts == 0:
    print("Account Locked due to too many failed attempts.")

# 25. ATM Menu
balance = 1000.0

while True:
    print("\n--- ATM Menu ---")
    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit")
    choice = input("Choose an option (1-4): ")

    if choice == '1':
        print(f"Current Balance: ${balance:.2f}")
    elif choice == '2':
        amount = float(input("Enter deposit amount: "))
        if amount > 0:
            balance += amount
            print(f"${amount:.2f} deposited successfully.")
        else:
            print("Invalid amount!")
    elif choice == '3':
        amount = float(input("Enter withdrawal amount: "))
        if 0 < amount <= balance:
            balance -= amount
            print(f"${amount:.2f} withdrawn successfully.")
        else:
            print("Insufficient balance or invalid amount!")
    elif choice == '4':
        print("Thank you for using the ATM. Goodbye!")
        break
    else:
        print("Invalid choice, please try again.")

# 26. Student Result System
name = input("Enter Student Name: ")
marks = []
for i in range(1, 4):
    m = float(input(f"Enter marks for subject {i} (out of 100): "))
    marks.append(m)

total = sum(marks)
percentage = total / 3

print("\n--- Student Result ---")
print(f"Name: {name}")
print(f"Total Marks: {total}/300")
print(f"Percentage: {percentage:.2f}%")

if percentage >= 40 and all(m >= 35 for m in marks):
    print("Result: PASS")
else:
    print("Result: FAIL")

# 27. Multiplication Table Generator
num = int(input("Enter the number for the table: "))
limit = int(input("Enter how many terms you want to print: "))

print(f"\nMultiplication Table of {num} up to {limit}:")
for i in range(1, limit + 1):
    print(f"{num} x {i} = {num * i}")
```
